```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet("../../data/EDA - SALES DATA/Sales/merged_data.parquet")
```

```python
df.head()
```

```python
df.info()
```

```python
null = df.columns[df.isnull().sum()>0].tolist()
null_dict = {i:df[i].isnull().sum() for i in null}
null_dict
```

```python
sorted_null = sorted(null_dict.items(), key=lambda x:x[1], reverse=True)
drop_null = [] # drop null values from col if null values is less than 1000 which is 1% of data
for i in range(len(sorted_null)):
    if sorted_null[i][1] < 1000:
        drop_null.append(sorted_null[i][0])
        
drop_null
```

```python
df = df.dropna(subset=drop_null, axis=0)
```

```python
df[df.columns[df.isnull().sum()>0]].isnull().sum()
```

```python
df.order_delivered_carrier_date = df.order_delivered_carrier_date.fillna(method='ffill', axis=0)

df.order_delivered_customer_date = df.order_delivered_customer_date.fillna(method='ffill', axis=0)

df.review_comment_title = df.review_comment_title.fillna('N/A', axis=0)

df.review_comment_message = df.review_comment_message.fillna('N/A', axis=0)
```

```python
len(np.where((df.product_category_name.isnull() ==True)&(df.product_name_lenght.isnull() ==True)&(df.product_description_lenght.isnull() ==True)&(df.product_photos_qty.isnull() ==True))[0])

# #all have same common null values
df.loc[df.index[np.where((df.product_category_name.isnull() ==True)&(df.product_name_lenght.isnull() ==True)&(df.product_description_lenght.isnull() ==True)&(df.product_photos_qty.isnull() ==True))[0]]]
# #can't work cuz all values are null

df.product_category_name.value_counts() #no product has been purchased single time wow!
# fas all 4 has same null we should fill them by "null_name"  ->> everything normal just it doesnot have basic name and lenght
df.product_category_name = df.product_category_name.fillna("Null", axis=0)
df.product_name_lenght = df.product_name_lenght.fillna("Null", axis=0)
df.product_description_lenght = df.product_description_lenght.fillna("Null", axis=0)
df.product_photos_qty = df.product_photos_qty.fillna("Null", axis=0)
df.isnull().sum().sum()
```
