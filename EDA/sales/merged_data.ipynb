```python
null = df.columns[df.isnull().sum()>0].tolist()
null_dict = {i:df[i].isnull().sum() for i in null}
null_dict
```

```python
sorted_null = sorted(null_dict.items(), key=lambda x:x[1], reverse=True)
drop_null = [] # drop null values from col if null values is less than 1000 which is 1% of data
for i in range(len(sorted_null)):
    if sorted_null[i][1] < 1000:
        drop_null.append(sorted_null[i][0])
        
drop_null
```
