```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_customers = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_customers_dataset.csv")
df_locations = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_geolocation_dataset.csv")
df_items = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_order_items_dataset.csv")
df_payments = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_order_payments_dataset.csv")
df_reviews = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_order_reviews_dataset.csv")
df_orders = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_orders_dataset.csv")
df_products = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_products_dataset.csv")
df_sellers = pd.read_csv("../../data/EDA - SALES DATA/Sales/olist_sellers_dataset.csv")
df_product_name = pd.read_csv("../../data/EDA - SALES DATA/Sales/product_category_name_translation.csv")

data = {'df_customers':df_customers,
        'df_locations':df_locations,
        'df_items':df_items,
        'df_payments':df_payments,
        'df_reviews':df_reviews,
        'df_orders':df_orders,
        'df_products':df_products,
        'df_sellers':df_sellers,
        'df_product_name':df_product_name}
```

```python
for df_name, df in data.items():
    print(df_name, ':', df.shape)
```

```python
df = df_orders.merge(df_items, on='order_id', how='left')

df = df.merge(df_payments, on='order_id', how='outer', validate='m:m')

df = df.merge(df_reviews, on='order_id', how='outer')

df = df.merge(df_products, on='product_id', how='outer')

df = df.merge(df_customers, on='customer_id', how='outer')

df = df.merge(df_sellers, on='seller_id', how='outer')

df.shape
```
